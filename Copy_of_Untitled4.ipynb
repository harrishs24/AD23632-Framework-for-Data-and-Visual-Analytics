```python
slider = widgets.IntSlider(value=50,min=0,max=100,step=1,description='Value:')
```

```python
output=widgets.Output()
def on_value_change(change):
  output.clear_output()
  with output:
    print(f'Slider value is{change["new"]}')

slider.observe(on_value_change,names = 'Values')

display(slider,output)
```

```python
import pandas as pd

import numpy as np

import seaborn as sns

import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler, StandardScaler

df = pd.read_csv('/content/fifa_world_cup_2026_player_performance.csv')

print("Initial Data Overview:")

print(df.info())
```

```python
print("\nMissing Values in Each Column:\n", df.isnull().sum())
sns.heatmap(df.isnull(), cbar=False, cmap="Blues")
plt.title("Missing Value Heatmap")

plt.show()

for col in ['player_name', 'team', 'jersey_number', 'position']:

  df[col].fillna(df[col].mode()[0])

df['total_minutes_tournament'].fillna(df['total_minutes_tournament'].median())
df['creativity_score'].fillna(df['creativity_score'].mode()[0])
df['clutch_performance_score'].fillna(df['clutch_performance_score'].mode()[0])
```

```python
initial_rows = df.shape[0]

df.drop_duplicates(inplace=True)

print(f"\nRemoved {initial_rows - df.shape[0]} duplicate rows.")
```

```python
df[col] = df[col].str.strip().str.capitalize()
```

```python
df['jersey_number'] = df['jersey_number'].replace('3+', 3).fillna(0).astype(int)
```

```python
min_max_scaler = MinMaxScaler()

scale_cols = ['creativity_score', 'clutch_performance_score', 'total_minutes_tournament']
```

```python
df[scale_cols] = min_max_scaler.fit_transform(df[scale_cols])
scaler = StandardScaler()

df[['height_cm']] = scaler.fit_transform(df[['height_cm']])
```

```python
print("\nCleaned Data Summary:")

print(df.info())

print(df.head())
```

```python
import pandas as pd
df = pd.read_csv("/content/fifa_world_cup_2026_player_performance.csv")
print(df.head())
print(df.tail())
print(df.info())
print(df.describe())
print(df.columns)
```

```python
categorical_cols = ['consistency_score', 'total_goals_tournament', 'club_name',
'successful_crosses', 'jersey_number']
```

```python
for col in categorical_cols:
  unique_vals = df[col].unique()
  print(f"\nUnique values in '{col}': {unique_vals}")
```

```python
df['consistency_score'] = df['consistency_score'].replace('3+', 3)
df['consistency_score'] = pd.to_numeric(df['consistency_score'], errors='coerce')
print("\nValue counts in 'consistency_score' after conversion:")

print(df['consistency_score'].value_counts(dropna=False))
```

```python
cols = ['position', 'jersey_number', 'club_name']
for col in cols:
    print(f"\nStatistics for '{col}':")
    if df[col].dtype == 'object': # Categorical column
        print(f"Mode: {df[col].mode().values}")
        print(f"Value Counts:\n{df[col].value_counts().head()}") # Display top value counts
    else: # Numerical column
        print(f"Mean: {df[col].mean():.2f}")
        print(f"Median: {df[col].median():.2f}")
        print(f"Mode: {df[col].mode().values}")
        print(f"Range: {df[col].max() - df[col].min():.2f}")
        print(f"Variance: {df[col].var():.2f}")
        print(f"Standard Deviation: {df[col].std():.2f}")
```
