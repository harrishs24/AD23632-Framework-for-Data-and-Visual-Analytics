## ABSTRACT

Tourism is one of the largest and fastest-growing sectors of the global economy, and decisions made by travel agencies, tourism boards and travellers all depend on reliable data. Raw tourism data, however, is usually scattered across many factors such as tourist arrivals, travel costs, ratings, seasons, festivals and visa rules, and it often contains duplicates and missing values.

This mini project presents an end-to-end analytics workflow on a global tourism dataset of **112 raw records** covering **25 destinations** across **18 countries** and **6 continents** for **2022–2025**. ### 8.3 Data Inspection
Before cleaning, the structure, data types, missing values and duplicate records are examined.

````python
print("=== STRUCTURE ===")
df.info()

print("\n=== MISSING VALUES PER COLUMN ===")
missing = df.isnull().sum()
print(missing[missing > 0] if missing.sum() else "No missing values")

print("\n=== DUPLICATE ROWS ===")
dup_count = df.duplicated().sum()
print("Duplicate records:", dup_count)

print("\n=== SUMMARY STATISTICS ===")
df.describe().round(2)
````

### 8.4 Data Cleaning
Two cleaning operations are applied:

1. **Remove duplicate records** so that no observation is counted twice.
2. **Median imputation** for the missing `Avg_Travel_Cost_USD` and `Tourist_Rating` values.

*Result reported in the original project:* 9 duplicates removed; final dataset = **103 rows × 18 columns** (including the three derived columns).

````python
rows_before = len(df)

# 1. Remove duplicates
df = df.drop_duplicates().reset_index(drop=True)
print(f"Duplicates removed : {rows_before - len(df)}")

# 2. Median imputation
for col in ["Tourist_Rating", "Avg_Travel_Cost_USD"]:
    n_missing = df[col].isnull().sum()
    med = df[col].median()
    df[col] = df[col].fillna(med)
    print(f"{col:<22}: filled {n_missing} missing value(s) with median = {med:.2f}")

print("\nRemaining missing values:", int(df.isnull().sum().sum()))
print("Rows after cleaning    :", len(df))
````

### 8.5 Feature Engineering (Derived Columns)

| Derived column | Logic |
|---|---|
| `Total_Travel_Cost_USD` | Accommodation + Food + Flight cost |
| `Cost_Category` | Budget / Moderate / Premium — the total cost is split into three equal-sized groups (terciles) |
| `Tourism_Growth_%` | Year-over-year percentage change in arrivals for each destination |

````python
# Total travel cost
df["Total_Travel_Cost_USD"] = (
    df["Accommodation_Cost_USD"] + df["Food_Cost_USD"] + df["Flight_Cost_USD"]
)

# Cost category (terciles of total cost)
df["Cost_Category"] = pd.qcut(
    df["Total_Travel_Cost_USD"], q=3, labels=["Budget", "Moderate", "Premium"]
)

# Year-over-year growth for every destination
df = df.sort_values(["Destination", "Year"]).reset_index(drop=True)
df["Tourism_Growth_%"] = (
    df.groupby("Destination")["Tourist_Arrivals"].pct_change() * 100
).round(2)
df["Tourism_Growth_%"] = df["Tourism_Growth_%"].fillna(0)   # first year has no previous year

print("Final shape:", df.shape)
df[["Destination", "Year", "Tourist_Arrivals", "Total_Travel_Cost_USD",
    "Cost_Category", "Tourism_Growth_%"]].head(8)
````

## 10. RESULTS AND KEY INSIGHTS

The code cell below recomputes the headline findings directly from the cleaned data.

````python
cont = df.groupby("Continent")["Tourist_Arrivals"].sum()
yr   = df.groupby("Year")["Tourist_Arrivals"].sum()
typ  = df.groupby("Destination_Type")["Avg_Travel_Cost_USD"].mean()
typ_arr = df.groupby("Destination_Type")["Tourist_Arrivals"].sum()
ctry = df.groupby("Country")["Tourist_Arrivals"].sum()
fest = df.groupby("Major_Festival")["Tourist_Arrivals"].sum()
r    = df["Avg_Travel_Cost_USD"].corr(df["Tourist_Rating"])

print("KEY FINDINGS FROM THE CLEANED DATASET")
print("-" * 60)
print(f"1. Total tourist arrivals            : {df['Tourist_Arrivals'].sum()/1e9:.2f} billion")
print(f"2. Average travel cost               : ${df['Avg_Travel_Cost_USD'].mean():,.0f}")
print(f"3. Average tourist rating            : {df['Tourist_Rating'].mean():.2f}")
print(f"4. Continent with most arrivals      : {cont.idxmax()}")
print(f"5. Country with most arrivals        : {ctry.idxmax()}")
print(f"6. Arrivals rose every year?         : {bool(yr.is_monotonic_increasing)}  ({yr.iloc[0]/1e6:.0f}M -> {yr.iloc[-1]/1e6:.0f}M)")
print(f"7. Destination type with most arrivals: {typ_arr.idxmax()}")
print(f"8. Costliest destination type        : {typ.idxmax()} (${typ.max():,.0f})")
print(f"9. Festival with most arrivals       : {fest.idxmax()}")
print(f"10. Cost vs rating correlation       : r = {r:.2f}")
````
